```sql
--Summary for which genres are rated best
CREATE OR REPLACE TABLE workspace.mini_etl.gold_genre_summary AS 
SELECT genre,
    COUNT(*)                 AS movie_count,
    ROUND(AVG(rating), 2)    AS avg_rating,
    MAX(rating)              AS highest_rating,
    current_timestamp()      AS loaded_at
FROM workspace.mini_etl.silver_movies
GROUP BY genre;
```
